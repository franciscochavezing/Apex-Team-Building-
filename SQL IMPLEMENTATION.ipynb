```python
try:
    connection = mysql.connector.connect(
        host="localhost",
        user="root",
        password="Pacho961101+",
        database="apex"
    )
    print("Connected to MySQL successfully")
    total_query = """
        SELECT COUNT(*) AS TotalFrequency
        FROM Player
    """
    cursor = connection.cursor()
    cursor.execute(total_query)
    total_result = cursor.fetchone()
    total_frequency = total_result[0]
    query = """
        SELECT 
            LegendID,
            COUNT(*) AS Frequency
        FROM 
            Player
        GROUP BY 
            LegendID
        ORDER BY 
            Frequency DESC
        ;
    """
    cursor.execute(query)
    rows = cursor.fetchall()
    cursor.close()
    connection.close()
    print("Connection closed")
    df = pd.DataFrame(rows, columns=['LegendID', 'Frequency'])
    df['Percentage'] = (df['Frequency'] / total_frequency) * 100
    print(df)
    plt.figure(figsize=(8, 8))
    plt.pie(df['Percentage'], labels=df['LegendID'], autopct='%1.1f%%', startangle=140)
    plt.title('Percentage of Usage of Legends',y=1.05)
    plt.axis('equal')  
    plt.show()

except mysql.connector.Error as e:
    print("Error connecting to MySQL:", e)
```

```python
try:
    connection = mysql.connector.connect(
        host="localhost",
        user="root",
        password="Pacho961101+",
        database="apex"
    )
    print("Connected to MySQL successfully")
    query = """
        SELECT 
            Finish_time_h - Start_time_h + (Finish_time_m - Start_time_m) / 60 AS GameDuration,
            COUNT(*) AS NumSquads
        FROM 
            Game
        INNER JOIN 
            Play ON Game.GameSerial = Play.GameSerial
        GROUP BY 
            Game.GameSerial;
    """
    cursor = connection.cursor()
    cursor.execute(query)
    rows = cursor.fetchall()
    cursor.close()
    connection.close()
    print("Connection closed")
    df = pd.DataFrame(rows, columns=['GameDuration', 'NumSquads'])
    df = df[(df['GameDuration'] > 0) & (df['NumSquads'] > 1)]
    plt.figure(figsize=(10, 6))
    plt.scatter(df['GameDuration'], df['NumSquads'], color='red', alpha=0.7)
    plt.xlabel('Game Duration (hours)')
    plt.ylabel('Number of Squads')
    plt.title('Relationship between Game Duration and Number of Squads ')
    plt.grid(True)
    plt.show()

except mysql.connector.Error as e:
    print("Error connecting to MySQL:", e)
```
